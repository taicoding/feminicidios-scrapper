```python
import os
from mongodb_client import MongoDBClient
mongo_uri = os.getenv("MONGO_URI")
db_name = os.getenv("DB_NAME")
collection = os.getenv("COLLECTION_NAME")
```

```python
client = MongoDBClient(mongo_uri, db_name, collection)
```

```python
client.connect()
```

```python
query = {"source": "eldiario", "tags": []}
documents = client.find_documents(query)
```

```python
len(documents)
```

```python
documents["tags"] = documents['section'].apply(lambda x: [t.lower().strip() for t in x.split("-")])
```

```python
documents["section"] = documents['tags'].apply(lambda x: x[1] if len(x)> 1 else x[0])
```

```python
for doc in documents.to_dict(orient='records'):
    query = {"url": doc["url"]}
    update = {"$set": {"tags": doc["tags"], "section": doc["section"]}}
    print(query, update)
    client.update_document(query, update)
```

```python
pipeline = {"source": "eldiario", "tags": {"$nin":["seguridad"] }}
delete_docs =  client.find_documents(pipeline)
```

```python
for doc in delete_docs.to_dict(orient='records'):
    query = {"url": doc["url"]}
    client.delete_document(query)
```
